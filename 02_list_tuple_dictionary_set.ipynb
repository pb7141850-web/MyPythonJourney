**Dictionary**: it is a collection of key-value pairs.
1. it is unordered
2. it is mutable
3. it is indexed
4. cannot contain duplicate keys

```python
a={
    "key":"value",
    "name":"Pooja",
    "cgpa":"8.9"
}
print(a)
print(a["name"])
```

**Dictionary methods**:
1. .items()  # returns a list of (key,value)tuple
2. .keys() # return a list containing dictionary keys
3. .update({"friends":}) # updates the dictionary with supplied key-value pairs
4. .get("name") # return the value of the keys
5. .clear() # remove all the elements from the dictionary
6. .copy() # return the shallow copy of the dictionary
7. .pop(key['key','default']) # removes the specified key and returns the corresponding value.If the key is not found ,'default' is returned if possible , otherwise 'keyword error' is raised
8. .popitem() # remove and return a key-value pair from the dictionary . Pairs are returned in LIFO

```python
a={
    "key":"value",
    "name":"Pooja",
    "cgpa":"8.9"
}
print(a.items())
print(a.keys())
a.update({"friends":"sachin"})
print(a)
print(a.get("name"))
print(a["name"])
print(a.get("name2")) # return None
print(a["name2"]) # raises the error
```

**SET IN PYTHON**: set is a collection of non-repititive elements.
1. unordered
2. unindexed
3. no duplication

```python
s={}  # empty dictionary
print(type(s))
e=set() # empty set
print(type(e))
```

**Methods in the SET**:
1. .add() # add the element in the set
2. len(set1) # return the length of the set
2. .clear() # remove all the element from the set
3. copy() # return a shallow copy of the set
4. .difference(set1.difference(set2)) # returns  a new set with elements in the set that are not in the other set
5. set1.difference_update(set2) # remove elements from the set that are also in the other set(s)
6. .discard(element) # remove an element from the set if it is a member else do nothing.
7. .intersection(others) # returns a new set elements common to the set and all other sets
8. .union({8,11}) # return a new set with all items from both sets
9. .remove(8) # updates the set s and removes 8 from s

```python
# 1. Defining the requested sets
set1 = {1, 2, 34, 34, 45, 5, 67}
set2 = {98, 67, 45, 23, 4}
print("set1:", set1)
print("set2:", set2)

# 2. len()
print("Length of set1:", len(set1))

# 3. .union()
print("Union:", set1.union(set2))

# 4. .intersection()
print("Intersection:", set1.intersection(set2))

# 5. .difference()
print("Difference (set1 - set2):", set1.difference(set2))
print("Difference (set2 - set1):", set2.difference(set1))

# 6. Subset and Superset check
subset_test = {1, 2}
print(f"{subset_test} is subset of set1?:", subset_test.issubset(set1))
print(f"set1 is superset of {subset_test}?:", set1.issuperset(subset_test))

# 7. .copy()
set_copy = set1.copy()
print("Copied set:", set_copy)

# 8. .add()
set_copy.add(100)
print("After adding 100:", set_copy)

# 9. .discard()
set_copy.discard(100)
set_copy.discard(999) # Does nothing as 999 is not in set_copy
print("After discarding 100:", set_copy)

# 10. .remove()
set_copy.remove(34)
print("After removing 34:", set_copy)

# 11. .difference_update()
set_copy.difference_update(set2)
print("After difference_update with set2:", set_copy)

# 12. .clear()
set_copy.clear()
print("After clearing copy:", set_copy)
```
