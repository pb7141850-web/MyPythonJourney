**If- else and elif**

```python
# if-elif-else ladder
a = int(input("Enter your age: "))
if(a>=18):    # identation is the space
  print("You can vote.")
elif(a<0):
  print("Age is not valid!")
else:
  print("You can't vote.")
```

```python
a = int(input("Enter your age: "))         # if can be independent but elifand else can't
# if statement no 1
if(a%2==0):
  print("a is even age.")
  # end of statement no 1
if(a>=18):     # if statement no 1
  print("You can vote.")
elif(a<0):
  print("Age is not valid!")
else:
  print("You can't vote.")
  # end of statement no 2
```
